```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
```

```python
plt.plot(x, y, color='green', linestyle='--', marker='o')
plt.title('My Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```
