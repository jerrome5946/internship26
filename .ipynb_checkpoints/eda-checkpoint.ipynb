```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# 1. Load the raw data and build the DataFrame 'df'
iris = load_iris()
df = pd.DataFrame(data=iris['data'], columns=iris['feature_names'])

# 2. Set up the canvas for a 2x2 grid of subplots
plt.figure(figsize=(12, 10))
plt.suptitle('Distribution of Iris Flower Measurements', fontsize=16, fontweight='bold')

# --- Plot 1: Sepal Length (Top-Left) ---
plt.subplot(2, 2, 1)  
plt.hist(df['sepal length (cm)'], edgecolor='black', color='skyblue')
plt.title('Sepal Length Distribution')
plt.xlabel('Length (cm)')
plt.ylabel('Count')

# --- Plot 2: Sepal Width (Top-Right) ---
plt.subplot(2, 2, 2)  
plt.hist(df['sepal width (cm)'], edgecolor='black', color='salmon')
plt.title('Sepal Width Distribution')
plt.xlabel('Width (cm)')
plt.ylabel('Count')

# --- Plot 3: Petal Length (Bottom-Left) ---
plt.subplot(2, 2, 3)  
plt.hist(df['petal length (cm)'], edgecolor='black', color='lightgreen')
plt.title('Petal Length Distribution')
plt.xlabel('Length (cm)')
plt.ylabel('Count')

# --- Plot 4: Petal Width (Bottom-Right) ---
plt.subplot(2, 2, 4)  
plt.hist(df['petal width (cm)'], edgecolor='black', color='gold')
plt.title('Petal Width Distribution')
plt.xlabel('Width (cm)')
plt.ylabel('Count')

# 3. Clean up spacing and display the final grid
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. Rebuild the standard DataFrame to avoid any missing variable errors
iris = load_iris()
df = pd.DataFrame(data=iris['data'], columns=iris['feature_names'])
df['species'] = iris['target']
df['species'] = df['species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Set up the plotting canvas
plt.figure(figsize=(10, 8))

# Get our list of unique species names
species = df['species'].unique()
colors = ['#FF5733', '#2ECC71', '#3498DB']  # Vibrant modern hex colors

# --- THE FOR LOOP (With Correct Indentation) ---
for i in range(len(species)):
    # Everything inside this loop is indented by 4 spaces (1 Tab)
    sub_df = df[df['species'] == species[i]]
    plt.scatter(
        sub_df['sepal length (cm)'], 
        sub_df['sepal width (cm)'], 
        c=colors[i], 
        label=species[i],
        s=70,          # Increases the point size so they are easier to see
        alpha=0.8      # Makes points slightly translucent to spot overlapping dots
    )

# 3. Add formatting context
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.title('Sepal Length vs Width by Species', fontsize=14, fontweight='bold')
plt.legend(title="Species Types")

plt.grid(True, linestyle=':', alpha=0.6) # Adds a clean background grid
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. Rebuild your specific dataframe fresh in memory
raw_data = load_iris()
iris_df = pd.DataFrame(
    raw_data.data, 
    columns=['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']
)
# Add the text column for species
iris_df['Species'] = raw_data.target
iris_df['Species'] = iris_df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Calculate correlation - dropping the text 'Species' column completely
# Fixed variable name to 'iris_df' and column key to 'Species'
corr = iris_df.drop('Species', axis=1).corr()

# 3. Plot the heatmap using your settings
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt=".2f", linewidths=0.5)

plt.title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. Rebuild your specific dataframe fresh in memory
raw_data = load_iris()
iris_df = pd.DataFrame(
    raw_data.data, 
    columns=['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']
)
# Add and map the Species column to text names
iris_df['Species'] = raw_data.target
iris_df['Species'] = iris_df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Set up the 2x2 subplot canvas grid
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# --- Top-Left: Sepal Length ---
sns.boxplot(x='Species', y='Sepal.Length', data=iris_df, ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Sepal Length by Species', fontsize=12, fontweight='bold')

# --- Top-Right: Sepal Width ---
sns.boxplot(x='Species', y='Sepal.Width', data=iris_df, ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Sepal Width by Species', fontsize=12, fontweight='bold')

# --- Bottom-Left: Petal Length ---
sns.boxplot(x='Species', y='Petal.Length', data=iris_df, ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Petal Length by Species', fontsize=12, fontweight='bold')

# --- Bottom-Right: Petal Width ---
sns.boxplot(x='Species', y='Petal.Width', data=iris_df, ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Petal Width by Species', fontsize=12, fontweight='bold')

# 3. Optimize layout and render
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# 1. Rebuild your specific dataframe fresh in memory
raw_data = load_iris()
iris_df = pd.DataFrame(
    raw_data.data, 
    columns=['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']
)
# Add and map the Species column to text names
iris_df['Species'] = raw_data.target
iris_df['Species'] = iris_df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Plot the violin chart using the correct keys
plt.figure(figsize=(12, 6))
sns.violinplot(x='Species', y='Petal.Length', data=iris_df, inner='quartile', palette='pastel')

plt.title('Petal Length Distribution by Species', fontsize=14, fontweight='bold')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates
from sklearn.datasets import load_iris

# 1. Rebuild your specific dataframe fresh in memory
raw_data = load_iris()
iris_df = pd.DataFrame(
    raw_data.data, 
    columns=['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']
)
# Add and map the Species column to text names
iris_df['Species'] = raw_data.target
iris_df['Species'] = iris_df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Plot the Parallel Coordinates chart
plt.figure(figsize=(12, 6))

# Fixed variable name to iris_df and column target to 'Species'
parallel_coordinates(iris_df, 'Species', colormap='viridis', linewidth=1.5, alpha=0.7)

plt.title('Parallel Coordinates Plot for Iris Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Flower Features')
plt.ylabel('Measurements (cm)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import andrews_curves
from sklearn.datasets import load_iris

# 1. Rebuild your specific dataframe fresh in memory
raw_data = load_iris()
iris_df = pd.DataFrame(
    raw_data.data, 
    columns=['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']
)
# Add and map the Species column to text names
iris_df['Species'] = raw_data.target
iris_df['Species'] = iris_df['Species'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

# 2. Plot the Andrews Curves using 'iris_df'
plt.figure(figsize=(10, 6))

# Fixed variable name to iris_df and column target to 'Species'
andrews_curves(iris_df, 'Species', colormap='viridis', linewidth=1.5, alpha=0.7)

# 3. Add title and layout formatting
plt.title('Andrews Curves for Iris Dataset', fontsize=14, fontweight='bold')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```
