```python
import pandas as pd

# 1. Add 'Gender' directly to your raw data
data = {
    'Name': ['Alice', 'Bob'], 
    'Age': [25, 30],
    'Gender': ['F', 'M']  # <-- Added here
}

df = pd.DataFrame(data)

# 2. Save it to the CSV
df.to_csv('output.csv', index=False)
print("CSV file created with Gender column!")
```
